## 1. 클래스(Class)란?

클래스는 **객체를 만들기 위한 설계도(청사진)** 입니다.

예를 들어, '팀원'이라는 개념을 클래스로 표현하면:
- **속성(Attribute)**: 이름, 직책, 연봉 등 → 팀원이 가지는 '데이터'
- **메서드(Method)**: 자기소개, 업무수행 등 → 팀원이 할 수 있는 '행동'

```
클래스 = 설계도
인스턴스(객체) = 설계도로 만들어진 실제 물건
```

## 3. `__init__` 메서드 (생성자)

- `__init__`은 인스턴스가 생성될 때 **자동으로 실행**되는 특별한 메서드입니다.
- `self`는 '나 자신(인스턴스)'을 가리키는 매개변수입니다.
- 인스턴스의 초기 속성(데이터)을 설정할 때 사용합니다.
- Docstring 매우 중요 => 향후 클래스 / 함수의 도움말 (메뉴얼) (문서화)

````python
# TeamMember 클래스 - 속성(데이터) 추가
class TeamMember:
    """ (Docstring)
    IT 회사 팀원을 표현하는 클래스
    
    속성:
        name (str): 팀원 이름
        role (str): 직책 (예: 시니어 개발자, 주니어 기획자 등)
        salary (int): 연봉 (만원 단위)
    """
    
    def __init__(self,name,role,salary):
        # self.속성명 = 값 => 인스턴스의 속성을 설정
        self.name = name
        self.role = role
        self.salary = salary
````

````python
alice = TeamMember(name="Alice", role="개발자", salary=3000)
bob = TeamMember(name="Bob", role="기획자", salary=3000)

# 속성에 접근 : 인스턴스.속성명
print(f"{alice.name}의 직책: {alice.role}, 연봉 : {alice.salary}만원")
print(f"{bob.name}의 직책: {bob.role}, 연봉 : {bob.salary}만원")
````

## 4. 인스턴스 메서드 추가하기

메서드는 클래스 내부에 정의된 함수입니다.  
첫 번째 매개변수는 항상 `self`(자기 자신)입니다.

````python
class TeamMember:
    """ (Docstring)
    IT 회사 팀원을 표현하는 클래스
    
    속성:
        name (str): 팀원 이름
        role (str): 직책 (예: 시니어 개발자, 주니어 기획자 등)
        salary (int): 연봉 (만원 단위)
    """
    
    def __init__(self,name,role,salary):
        # self.속성명 = 값 => 인스턴스의 속성을 설정
        self.name = name
        self.role = role
        self.salary = salary
        self.projects = []    # 현재 참여 중인 프로젝트 목록

        self.completed_projects = []  # 수행 완료한 프로젝트 목록
        self.is_at_work = False       # 출근 여부 (True: 출근, False: 퇴근/미출근)
        self.is_resting = False       # 휴식 여부 (True: 휴식 중, False: 업무 중)
    
    # 메서드 정의
    def introduce(self):
        """자기소개 메서드"""
        print(f"안녕하세요! 저는 {self.role} {self.name} 입니다.")

    # 1. 프로젝트를 부여하는 행동
    def assign_project(self, project_name):
        if project_name in self.projects:
            print(f"[{self.name}] 이미 '{project_name}' 프로젝트에 참여 중입니다.")
        else:
            self.projects.append(project_name)
            print(f"[{self.name}] 새로운 프로젝트 '{project_name}'을(를) 부여받았습니다.")
            
    # 2. 프로젝트를 수행(완료)하는 행동
    def complete_project(self, project_name):
        if project_name in self.projects:
            self.projects.remove(project_name)
            self.completed_projects.append(project_name)
            print(f"[{self.name}] '{project_name}' 프로젝트를 성공적으로 완수했습니다!")
        else:
            print(f"[{self.name}] 진행 중인 프로젝트 중 '{project_name}'은(는) 없습니다.")

    # 3. 연봉을 바꾸는 행동 (연봉 인상/삭감 고려)
    def update_salary(self, new_salary):
        if new_salary <= 0:
            print("연봉은 0원 이하가 될 수 없습니다.")
            return
        
        diff = new_salary - self.salary
        self.salary = new_salary
        
        if diff > 0:
            print(f"[{self.name}] 축하합니다! 연봉이 {diff}만원 인상되어 {self.salary}만원이 되었습니다.")
        elif diff < 0:
            print(f"[{self.name}] 연봉이 {abs(diff)}만원 삭감되어 {self.salary}만원이 되었습니다.")
        else:
            print(f"[{self.name}] 연봉이 동결되었습니다 ({self.salary}만원).")

    # 4. 직책을 바꾸는 행동 (진급 등)
    def change_role(self, new_role):
        old_role = self.role
        self.role = new_role
        print(f"[{self.name}] 직책이 {old_role} -> {new_role}(으)로 변경되었습니다.")

    # 5. 출근을 하는 행동
    def work_in(self):
        if self.is_at_work:
            print(f"[{self.name}] 이미 출근한 상태입니다.")
        else:
            self.is_at_work = True
            print(f"[{self.name}] 오늘 대기 상태: 출근 완료했습니다. 좋은 하루 되세요!")

    # 6. 퇴근을 하는 행동
    def work_out(self):
        if not self.is_at_work:
            print(f"[{self.name}] 아직 출근하지 않았거나 이미 퇴근했습니다.")
        else:
            # 퇴근 시 휴식 상태도 함께 초기화
            self.is_at_work = False
            self.is_resting = False
            print(f"[{self.name}] 퇴근합니다. 고생하셨습니다!")

    # 7. 휴식을 하는 행동
    def take_a_break(self):
        if not self.is_at_work:
            print(f"[{self.name}] 출근 상태가 아니라서 휴식할 수 없습니다.")
            return
        
        if self.is_resting:
            print(f"[{self.name}] 이미 휴식 중입니다.")
        else:
            self.is_resting = True
            print(f"[{self.name}] 커피 브레이크! 현재 휴식 중입니다.")

    # 8. 업무로 복귀하는 행동 (휴식 종료)
    def resume_work(self):
        if not self.is_resting:
            print(f"[{self.name}] 이미 업무 중이거나 휴식 상태가 아닙니다.")
        else:
            self.is_resting = False
            print(f"[{self.name}] 휴식을 마치고 업무로 복귀합니다.")
````

````python
# 1. 객체 생성 (기존에 정의된 클래스를 바탕으로 생성)
elice = TeamMember(name="elice", role="개발자", salary=3000)

print("=== 1. 기본 소개 ===")
elice.introduce()
print()

print("=== 2. 출근 전 행동 제어 (예외 처리 테스트) ===")
# 출근하지 않은 상태에서 휴식을 시도할 때의 예외 처리를 확인합니다.
elice.take_a_break() 
print()

print("=== 3. 정식 출근 및 업무 시작 ===")
elice.work_in()
# 이미 출근했는데 또 출근을 시도할 경우
elice.work_in() 
print()

print("=== 4. 프로젝트 부여 및 수행 ===")
# 프로젝트 할당
elice.assign_project("AI 챗봇 서비스 구축")
elice.assign_project("신규 웹사이트 리뉴얼")
# 동일한 프로젝트 중복 부여 시도
elice.assign_project("AI 챗봇 서비스 구축") 

print(f"현재 진행 중인 프로젝트: {elice.projects}")
print()

print("=== 5. 업무 중 휴식 ===")
elice.take_a_break()     # 휴식 시작
elice.resume_work()      # 업무 복귀
print()

print("=== 6. 프로젝트 완수 ===")
# 부여받은 프로젝트 중 하나를 완료 처리
elice.complete_project("AI 챗봇 서비스 구축")

print(f"남은 프로젝트: {elice.projects}")
print(f"완료한 프로젝트 이력: {elice.completed_projects}")
print()

print("=== 7. 성과 인정: 직책 변경 및 연봉 협상 ===")
# 뛰어난 성과로 인한 진급 및 연봉 인상 시나리오
elice.change_role("시니어 개발자")
elice.update_salary(4500)  # 3000만 원에서 4500만 원으로 인상
print()

print("=== 8. 하루 업무 마감 및 퇴근 ===")
elice.work_out()
# 이미 퇴근했는데 또 퇴근을 시도할 경우
elice.work_out()
````

## 5. 클래스 변수(Class Variable) vs 인스턴스 변수(Instance Variable)

| 구분 | 설명 | 선언 위치 |
|------|------|----------|
| **클래스 변수** | 모든 인스턴스가 **공유**하는 변수 | 클래스 내부, `__init__` 바깥 |
| **인스턴스 변수** | 각 인스턴스마다 **독립적**으로 가지는 변수 | `__init__` 내부, `self.변수명` |

````python
class TeamMember:
    # - 클래스 변수 : 모든 팀원이 같은 값을 공유 -
    company_name = "엘리스"    # 회사 이름은 모두 동일
    total_members = 0          # 회사 직원 수

    def __init__(self,name,role,salary):
        # - 인스턴스 변수 : 각 티원마다 다른 값 - 
        self.name = name
        self.role = role
        self.salary = salary
        
        # 인스턴스 생성될 때 마다 클래스 변수 업데이트 
        # 새로운 팀원이 생성될 때 마다 회사 직원 수는 증가한다.
        TeamMember.total_members += 1       
        
    def introduce(self):
        print(f"[{self.company_name}] {self.role} {self.name} (연봉 : {self.salary}만원)")
doran = TeamMember(name="Doran", role="개발자", salary=3000)
faker = TeamMember(name="Faker", role="기획자", salary=3000)

doran.introduce()
faker.introduce()

# 팀원 수
print(f"\n 회사 총 직원 수 : {TeamMember.total_members} 명")
print(f"\n 회사 총 직원 수 : {faker.total_members} 명")

print("===추가===")
TeamMember.total_members = TeamMember.total_members + 1
faker.total_members=10

# 이건 클래스 변수 라고 생각이 되고
print(f"\n 회사 총 직원 수 : {TeamMember.total_members} 명")  

# 이건 인스턴스 변수 인가?
# 인스턴스에서 이 변수를 접근해서 수정을 하면 클래스 변수 -> 인스턴스 변수로 변한 것인지?
# 아니면 인스턴스 생성 했을 때, 이 변수가 똑같이 생성이 되면서 인스턴스 변수로 선언이 된것인지?
print(f"\n 회사 총 직원 수 : {faker.total_members} 명")       

#
# TeamMember.total_members는 클래스 변수, 모든 인스턴스가 이 하나의 메모리 공간을 공유 한다.
# faker.total_members는 인스턴스 생성 시 존재하지 않는 변수이고
# faker.total_members = 10 이 순간, faker 객체 전용 인스턴스 변수가 새로 생성된다. -> 클래스 변수는 변하지 않는다.
# 
# Python에서 `인스턴스.변수명` 호출 시 단계
# 1. 해당 인스턴스 안에 변수가 있는지 찾는다.
# 2. 없으면 인스턴스를 만든 클래스로 가서 변수를 찾는다.
````

## 7. 클래스 메서드(Class Method)와 정적 메서드(Static Method)

| 종류 | 데코레이터 | 첫 번째 매개변수 | 용도 |
|------|-----------|-----------------|------|
| **인스턴스 메서드** | 없음 | `self` | 인스턴스 데이터 사용 |
| **클래스 메서드** | `@classmethod` | `cls` | 클래스 변수 사용, 대체 생성자 |
| **정적 메서드** | `@staticmethod` | 없음 | 클래스와 관련은 있지만 독립적인 기능 |

````python
# 코드 
class TeamMember:
    company_name = "TechCorp"
    total_members = 0
    
    def __init__(self, name, role, salary):
        self.name = name
        self.role = role
        self.salary = salary
        TeamMember.total_members +=1

    # - 클래스 메서드 : cls = 클래스 자체를 받음
    @classmethod       # @classmethod -> 클래스 메서드로 선언
    def get_total_members(cls):
        """현재까지 생성된 팀원 수 반환"""
        return cls.total_members

    @classmethod
    def from_string(cls, member_string):
        """
        '이름-직책-연봉' 형식의 문자열로 팀원을 생성하는 대체 생성자
        예) "Alice-시니어 개발자-6000"
        """
        name, role, salary = member_string.split("-")
        return cls(name, role, int(salary))   # cls() = TeamMember()

    # - 정적 메서드 : self 나 cls 없음
    @staticmethod
    def is_valid_salary(salary):
        """연봉이 유요한 범위인지 검사하는 유틸리티 함수"""
        return 1000 <= salary <= 10000

    def __str__(self):
        return f"{self.role} {self.name}"

    def __repr__(self):
        return f"TeamMember(name='{self.name}',role='{self.role}',salary='{self.salary}')"
````

````python
doran = TeamMember(name="Doran", role="개발자", salary=3000)
faker = TeamMember(name="Faker", role="기획자", salary=3000)

print(f"총 팀원 수 : {TeamMember.get_total_members()}명")
# 이 클래스 메서드로 객체 생성 한것 맞나? -> 맞음 대체 생성자
peyz = TeamMember.from_string("Peyz-데이터분석가-3300") 
print(f"영입 : {peyz}")
print(f"총 팀원 수 : {TeamMember.get_total_members()}명")

print(TeamMember.is_valid_salary(doran.salary))
TeamMember.is_valid_salary(500) # 이건 Jupyter Note 라서 출력 되는 것으로 보이는 것이지, 실제로 print 되진 않았음
````

## 8. 상속(Inheritance) – 팀(부서) 클래스 만들기

상속은 기존 클래스의 속성과 메서드를 물려받아 **새로운 클래스를 만드는 것**입니다.

```
Department (부서 기본 클래스)
    ├── DevTeam      (개발팀)
    ├── PlanningTeam (기획팀)
    ├── SalesTeam    (영업팀)
    └── AnalyticsTeam(분석팀)
```

````python
# 코드 
# 부모 클래스 (Bas Class)
# 모든 팀이 공통으로 가지는 속성과 행동
class Department:
    """
    IT 회사 부서(팀)를 표현하는 기본 클래스
    
    속성:
        team_name (str): 팀 이름
        members (list): 팀원 목록
        budget (int): 팀 예산 (만원)
    """
    
    def __init__(self, team_name, budget):
        self.team_name = team_name
        self.budget = budget
        self.members = []         # 팀원 목록 (TeamMember 인스턴스들)
    
    def add_member(self, member):
        """팀원을 팀에 추가"""
        self.members.append(member)
        print(f"[{self.team_name}] {member.name}님이 합류했습니다.")
    
    def remove_member(self, name):
        """이름으로 팀원을 제거"""
        for member in self.members:
            if member.name == name:
                self.members.remove(member)
                print(f"[{self.team_name}] {name}님이 퇴팀했습니다.")
                return
        print(f"'{name}'님을 찾을 수 없습니다.")
    
    def show_members(self):
        """팀 구성원 목록 출력"""
        print(f"\n=== {self.team_name} 구성원 ({len(self.members)}명) ===")
        if not self.members:
            print("  (팀원 없음)")
        for i, member in enumerate(self.members, 1):
            print(f"  {i}. {member.role}: {member.name} (연봉 {member.salary}만원)")
    
    def get_total_salary(self):
        """팀 전체 연봉 합계 반환"""
        return sum(m.salary for m in self.members)
    
    def team_meeting(self):
        """팀 회의 (자식 클래스에서 재정의 가능)"""
        print(f"[{self.team_name}] 정기 팀 회의를 시작합니다.")
    
    def __str__(self):
        return f"{self.team_name} (팀원 {len(self.members)}명, 예산 {self.budget}만원)"
````

````python
# ────────────────────────────────────────────
# 자식 클래스 1: 개발팀
# ────────────────────────────────────────────
class DevTeam(Department):
    """
    개발팀 클래스 - Department를 상속
    
    추가 속성:
        tech_stack (list): 사용 기술 스택
    """
    
    def __init__(self, budget, tech_stack):
        # super().__init__()으로 부모 클래스의 __init__ 호출
        super().__init__("개발팀", budget)
        self.tech_stack = tech_stack     # 개발팀만 가지는 추가 속성
    
    def code_review(self):
        """개발팀 전용 메서드: 코드 리뷰 진행"""
        print(f"[{self.team_name}] 코드 리뷰를 시작합니다. 참여자: {', '.join(m.name for m in self.members)}")
    
    def show_tech_stack(self):
        """사용 기술 스택 출력"""
        print(f"[{self.team_name}] 기술 스택: {', '.join(self.tech_stack)}")
    
    # 부모 메서드 오버라이딩(Override): 개발팀만의 회의 방식
    def team_meeting(self):
        print(f"[{self.team_name}] 스프린트 계획 회의를 시작합니다. (기술 스택: {', '.join(self.tech_stack)})")
````

````python
# ────────────────────────────────────────────
# 자식 클래스 2: 기획팀
# ────────────────────────────────────────────
class PlanningTeam(Department):
    """
    기획팀 클래스 - Department를 상속
    
    추가 속성:
        current_plan (str): 현재 기획 중인 서비스명
    """
    
    def __init__(self, budget):
        super().__init__("기획팀", budget)
        self.current_plan = None
    
    def start_planning(self, service_name):
        """새 서비스 기획 시작"""
        self.current_plan = service_name
        print(f"[{self.team_name}] '{service_name}' 서비스 기획을 시작합니다.")
    
    def team_meeting(self):
        plan_info = f"(현재 기획: {self.current_plan})" if self.current_plan else ""
        print(f"[{self.team_name}] 기획 검토 회의를 시작합니다. {plan_info}")
````

````python
# ────────────────────────────────────────────
# 자식 클래스 3: 영업팀
# ────────────────────────────────────────────
class SalesTeam(Department):
    """
    영업팀 클래스 - Department를 상속
    
    추가 속성:
        monthly_target (int): 월 매출 목표 (만원)
        monthly_achievement (int): 월 매출 달성액 (만원)
    """
    
    def __init__(self, budget, monthly_target):
        super().__init__("영업팀", budget)
        self.monthly_target = monthly_target
        self.monthly_achievement = 0
    
    def close_deal(self, amount):
        """계약 성사 시 매출 기록"""
        self.monthly_achievement += amount
        rate = (self.monthly_achievement / self.monthly_target) * 100
        print(f"[{self.team_name}] 계약 성사! +{amount}만원 (이번 달 달성률: {rate:.1f}%)")
    
    def team_meeting(self):
        rate = (self.monthly_achievement / self.monthly_target) * 100
        print(f"[{self.team_name}] 영업 현황 회의 (목표: {self.monthly_target}만원, 달성: {self.monthly_achievement}만원, {rate:.1f}%)")
````

````python
# ────────────────────────────────────────────
# 자식 클래스 4: 분석팀
# ────────────────────────────────────────────
class AnalyticsTeam(Department):
    """
    분석팀 클래스 - Department를 상속
    
    추가 속성:
        tools (list): 사용 분석 도구
        reports (list): 완료된 분석 보고서 목록
    """
    
    def __init__(self, budget, tools):
        super().__init__("분석팀", budget)
        self.tools = tools
        self.reports = []
    
    def publish_report(self, report_name):
        """분석 보고서 발행"""
        self.reports.append(report_name)
        print(f"[{self.team_name}] 보고서 발행: '{report_name}'")
    
    def team_meeting(self):
        print(f"[{self.team_name}] 데이터 리뷰 회의 (사용 도구: {', '.join(self.tools)}, 발행 보고서: {len(self.reports)}건)")


print("클래스 정의 완료!")
````

````python
# ── Step 6: 전체 팀 회의 ──
print("=" * 50)
print("Step 6. 전체 팀 회의 (다형성 - Polymorphism)")
print("=" * 50)

print("※ 같은 team_meeting() 메서드를 호출해도 팀마다 다르게 동작합니다.")

for team in [dev_team, planning_team, sales_team, analytics_team]:
    team.team_meeting()
````

## 11. 정리: 핵심 개념 요약

| 개념 | 설명 | 예시 |
|------|------|------|
| **클래스** | 객체의 설계도 | `class TeamMember:` |
| **인스턴스** | 클래스로 만든 실제 객체 | `alice = TeamMember(...)` |
| **`__init__`** | 인스턴스 생성 시 자동 실행 | 초기 속성 설정 |
| **인스턴스 변수** | 각 인스턴스마다 독립적인 데이터 | `self.name` |
| **클래스 변수** | 모든 인스턴스가 공유하는 데이터 | `total_members = 0` |
| **인스턴스 메서드** | `self`를 받는 일반 메서드 | `def introduce(self):` |
| **클래스 메서드** | `@classmethod`, `cls`를 받음 | 대체 생성자 등 |
| **정적 메서드** | `@staticmethod`, 독립적 기능 | 유틸리티 함수 |
| **상속** | 부모 클래스를 물려받아 확장 | `class DevTeam(Department):` |
| **오버라이딩** | 부모 메서드를 자식에서 재정의 | `team_meeting()` |
| **다형성** | 같은 메서드 호출, 다른 동작 | `for team in all_teams: team.team_meeting()` |
