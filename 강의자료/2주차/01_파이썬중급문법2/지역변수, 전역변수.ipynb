# 지역 변수 vs 전역변수
> 지역변수 (Local) : 함수 안에서 정의-> 함수 안에서만 존재 (밖에서 접근 불가)
 
> 전역변수 (Global) : 함수 밖에서 정의 -> 내부 프로그램 어디에서나 접근 가능
>


## 지역 변수
- 함수 안에서만 존재

```python
def greet():
    msg = "함수 안에서 안녕" # 지역변수
    print(msg)

greet()
```

## 전역 변수
- 어디서나 접근 가능

```python
msg = "python is awesome!" # 전역변수
def display():
    print(f"함수 안: {msg}")

display()
```

## Shadowing
- 같은 이름의 지역 변수가 전역변수를 가림

```python
s = "I love yeardreamschool"

def fun():
    s = "Me too."
    print(s)

fun()
print(s)
```

## 전역 변수를 수정하고 싶다면?
- `global` 필요

```python
count = 0                    # 전역변수
print(f"count : {count}")
def increase():
    global count             # global 선언이 있으면 전역을 바꿀 수 있다.
    count = count +1

increase()
print(f"count : {count}")    # 전역변수

print("반복문 5회 실시")
for i in range(5):
    increase()

print(f"count : {count}")     # 전역변수
```

```python
count = 0                     #전역변수 선언
def increase(count=0):        # 왜 파라미터를 count=0으로 하나?
    count = count + 1         # 지역변수 인거 같은데?

increase()
print(count)
```
